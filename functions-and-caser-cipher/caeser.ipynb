Caeser Cipher


```python
alphabet = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j', 'k', 'l', 'm', 'n', 'o', 'p', 'q', 'r', 's', 't', 'u', 'v', 'w', 'x', 'y', 'z']

def caesar(text,shift,action):
    required_text = ""
    new_index = 0
    for letter in text:
        if letter == " " or not letter.isalpha():
            required_text += letter
        else:
            get_index = alphabet.index(letter)
            if action == "encode":
                new_index = (get_index + shift) % 26
                required_text += alphabet[new_index]
            else:
                new_index = (get_index - shift) % 26
                required_text += alphabet[new_index]
    return required_text
```
