```python
from replit import clear
from art import logo
#HINT: You can call clear() to clear the output in the console.
print(logo)
bid_dictionary = {}
bidding_finished = False


#find highest bidder
def find_highest_bidder(bid_dictionary):
  highest_bid = 0
  winner = ""
  for bidder in bid_dictionary:
    bid_amount = bid_dictionary[bidder]
    if bid_amount > highest_bid:
      highest_bid = bid_amount
      winner = bidder
  print(f"The winner is {winner} with a bid of ${highest_bid}")
  print(bid_dictionary)


while not bidding_finished:
  name = input("what is your name?: ")
  price = int(input("what is your bid?: $"))
  bid_dictionary[name] = price
  answer = input("Are there any other bidders? Type 'yes or 'no'.\n").lower()
  if answer == "no":
    bidding_finished = True
    find_highest_bidder(bid_dictionary)
  else:
    clear()
```
